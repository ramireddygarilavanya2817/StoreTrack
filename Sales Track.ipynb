```python
# Product Catalogue

product_catalogue = {}

for index, row in df.drop_duplicates("ProductID").iterrows():

    product_id = row["ProductID"]

    product_catalogue[product_id] = {
        "ProductName": row["ProductName"],
        "Category": row["Category"],
        "UnitPrice": row["UnitPrice"],
        "GSTPercent": row["GSTPercent"],
        "StockQuantity": 20
    }

print("Product Catalogue Created Successfully!")
```

```python
# Shopping Cart

cart = []

print("Shopping cart created successfully!")
```

```python
def add_to_cart(product_id, quantity):

    # Check whether product exists
    product = None
    actual_id = None

    for pid in product_catalogue:
        if str(pid) == str(product_id):
            product = product_catalogue[pid]
            actual_id = pid
            break

    if product is None:
        print("Invalid Product ID!")
        return

    # Validate quantity
    try:
        quantity = int(quantity)
    except ValueError:
        print("Please enter a valid quantity!")
        return

    if quantity <= 0:
        print("Quantity must be greater than zero!")
        return

    # Check existing cart quantity
    existing_quantity = 0

    for item in cart:
        if item["ProductID"] == actual_id:
            existing_quantity += item["Quantity"]

    # Check stock availability
    available_stock = product["StockQuantity"]

    if existing_quantity + quantity > available_stock:
        print("Insufficient stock!")
        print("Available stock:", available_stock - existing_quantity)
        return

    # Add product to cart
    cart.append({
        "ProductID": actual_id,
        "ProductName": product["ProductName"],
        "UnitPrice": product["UnitPrice"],
        "Quantity": quantity,
        "GSTPercent": product["GSTPercent"]
    })

    print(product["ProductName"], "added to cart successfully!")
```

```python
subtotal = 0

for item in cart:
    subtotal += item["UnitPrice"] * item["Quantity"]

print("Cart Subtotal:", round(subtotal, 2))
```

```python
def calculate_discount(subtotal):

    if subtotal >= 2000:
        discount_percent = 15
    elif subtotal >= 1000:
        discount_percent = 10
    elif subtotal >= 500:
        discount_percent = 5
    else:
        discount_percent = 0

    discount_amount = subtotal * discount_percent / 100

    return discount_percent, discount_amount
```

```python
def calculate_gst(taxable_amount, gst_percent):

    gst_amount = taxable_amount * gst_percent / 100

    return gst_amount
```

```python
subtotal = sum(
    item["UnitPrice"] * item["Quantity"]
    for item in cart
)

discount_percent, discount_amount = calculate_discount(subtotal)

taxable_amount = subtotal - discount_amount

total_gst = 0

for item in cart:
    item_amount = item["UnitPrice"] * item["Quantity"]
    item_discount = item_amount * discount_percent / 100
    item_taxable = item_amount - item_discount

    total_gst += calculate_gst(
        item_taxable,
        item["GSTPercent"]
    )

final_amount = taxable_amount + total_gst

print("========== STORETRACK BILL ==========")
print("Subtotal:", round(subtotal, 2))
print("Discount:", discount_percent, "%")
print("Discount Amount:", round(discount_amount, 2))
print("Taxable Amount:", round(taxable_amount, 2))
print("GST Amount:", round(total_gst, 2))
print("Final Amount:", round(final_amount, 2))
print("=====================================")
```

```python
def checkout():

    if not cart:
        print("Cart is empty!")
        return

    payment_method = input("Enter payment method (Cash/UPI/Card): ").strip().upper()

    if payment_method not in ["CASH", "UPI", "CARD"]:
        print("Invalid payment method!")
        return

    subtotal = sum(
        item["UnitPrice"] * item["Quantity"]
        for item in cart
    )

    discount_percent, discount_amount = calculate_discount(subtotal)
    taxable_amount = subtotal - discount_amount

    total_gst = 0

    for item in cart:
        amount = item["UnitPrice"] * item["Quantity"]
        item_discount = amount * discount_percent / 100
        item_taxable = amount - item_discount

        total_gst += calculate_gst(
            item_taxable, item["GSTPercent"]
        )

    final_amount = taxable_amount + total_gst

    invoice_number = "INV" + datetime.now().strftime("%Y%m%d%H%M%S")

    filename = "sales_history.csv"
    file_exists = os.path.exists(filename)

    with open(filename, "a", newline="") as file:
        writer = csv.writer(file)

        if not file_exists:
            writer.writerow([
                "InvoiceNo", "Date", "CustomerID", "ProductID",
                "ProductName", "Category", "Quantity", "UnitPrice",
                "DiscountPercent", "GSTPercent", "Subtotal",
                "DiscountAmount", "TaxableAmount", "GSTAmount",
                "TotalAmount", "PaymentMethod"
            ])

        for item in cart:
            pid = item["ProductID"]
            product = product_catalogue[pid]

            amount = item["UnitPrice"] * item["Quantity"]
            item_discount = amount * discount_percent / 100
            item_taxable = amount - item_discount
            item_gst = calculate_gst(
                item_taxable, item["GSTPercent"]
            )

            writer.writerow([
                invoice_number,
                datetime.now().strftime("%Y-%m-%d"),
                "WALKIN",
                pid,
                item["ProductName"],
                product["Category"],
                item["Quantity"],
                item["UnitPrice"],
                discount_percent,
                item["GSTPercent"],
                amount,
                item_discount,
                item_taxable,
                item_gst,
                item_taxable + item_gst,
                payment_method
            ])

    # Update stock
    for item in cart:
        pid = str(item["ProductID"])
        stock_data[pid]["StockQuantity"] -= item["Quantity"]
        product_catalogue[item["ProductID"]]["StockQuantity"] = (
            stock_data[pid]["StockQuantity"]
        )

    with open("stock.json", "w") as file:
        json.dump(stock_data, file, indent=4)

    print("\n========== FINAL BILL ==========")
    print("Invoice Number:", invoice_number)
    print("Subtotal: ₹", round(subtotal, 2))
    print("Discount: ₹", round(discount_amount, 2))
    print("GST: ₹", round(total_gst, 2))
    print("Final Amount: ₹", round(final_amount, 2))
    print("Payment:", payment_method)
    print("================================")

    cart.clear()

    print("Sales saved successfully!")
    print("Stock updated successfully!")
    print("Cart cleared!")
```

```python
def storetrack_menu():

    while True:
        print("\n========== STORETRACK ==========")
        print("1. View Product Catalogue")
        print("2. Add Product to Cart")
        print("3. View Shopping Cart")
        print("4. Check Stock")
        print("5. View Sales History")
        print("6. Generate Bill / Checkout")
        print("7. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            for pid, details in product_catalogue.items():
                print(pid, "-", details["ProductName"],
                      "| Price:", details["UnitPrice"],
                      "| Stock:", details["StockQuantity"])

        elif choice == "2":
            product_id = input("Enter Product ID: ")
            quantity = input("Enter Quantity: ")
            add_to_cart(product_id, quantity)

        elif choice == "3":
            if not cart:
                print("Your cart is empty!")
            else:
                for item in cart:
                    print(item["ProductName"],
                          "| Quantity:", item["Quantity"],
                          "| Price:", item["UnitPrice"])

        elif choice == "4":
            for pid, details in product_catalogue.items():
                print(details["ProductName"],
                      "- Available:", details["StockQuantity"])

        elif choice == "5":
            if os.path.exists("sales_history.csv"):
                history = pd.read_csv("sales_history.csv")
                display(history.tail(10))
            else:
                print("No sales history found!")

        elif choice == "6":
            checkout()

        elif choice == "7":
            print("Thank you for using StoreTrack!")
            break

        else:
            print("Invalid choice! Please try again.")
```
